# CIE-III: Statistical Analysis, EDA, Feature Engineering, **Visualization**

```python
df['grade_improvement'] = df['grades_after_ai'] - df['grades_before_ai'] #Create new column
df.head()
```

### Statistical Analysis

```python
print("Average Improvement:", df['grade_improvement'].mean())
print("Maximum Improvement:", df['grade_improvement'].max())
print("Minimum Improvement:", df['grade_improvement'].min())
```

### EDA (Exploratory Data Analysis)

```python
df['uses_ai'].value_counts()
```

```python
df['education_level'].value_counts()
```

### Visualization

```python
plt.figure()
df['grade_improvement'].hist()
plt.title("Grade Improvement Distribution")
plt.xlabel("Improvement")
plt.ylabel("Frequency")
plt.show()
```

### Study Hours vs Grades After AI

```python
plt.figure()
plt.scatter(df['study_hours_per_day'], df['grades_after_ai'])
plt.title("Study Hours vs Grades After AI")
plt.xlabel("Study Hours")
plt.ylabel("Grades After AI")
plt.show()
```

### AI Users vs Non-AI Users

```python
df.groupby('uses_ai')['grade_improvement'].mean().plot(kind='bar')
plt.title("AI Usage vs Grade Improvement")
plt.xlabel("Uses AI")
plt.ylabel("Average Improvement")
plt.show()
```

### Correlation Heatmap

```python
import plotly.express as px

# 1. Group by AI usage and calculate mean
comparison_table = df.groupby('uses_ai').mean(numeric_only=True)

# 2. Remove 'age' column (if exists)
comparison_table = comparison_table.drop(columns=['age'], errors='ignore')

# 3. Transpose (features as rows)
comparison_table = comparison_table.T

# 4. Round values for clean display
comparison_table = comparison_table.round(2)

# 5. Plotly heatmap
fig = px.imshow(
    comparison_table,
    text_auto=True,
    color_continuous_scale='RdBu_r',
    aspect='auto',
    title="AI vs Non-AI Feature Comparison Heatmap"
)

fig.update_layout(
    xaxis_title="Uses AI",
    yaxis_title="Features",
    width=800,
    height=600
)

fig.show()
```

```python
import statistics as stats

# Sample dataset (exam scores)
data = [85, 90, 78, 92, 88, 76, 95, 89, 84, 91]

print("Dataset:", data)

# (1) Mean
mean_val = stats.mean(data)
print("(1) Mean:", mean_val)

# (2) Median
median_val = stats.median(data)
print("(2) Median:", median_val)

# (3) Mode
mode_val = stats.mode(data)
print("(3) Mode:", mode_val)

# (4) Range
range_val = max(data) - min(data)
print("(4) Range:", range_val)

# (5) Variance
variance_val = stats.variance(data)
print("(5) Variance:", variance_val)

# (6) Standard Deviation
std_dev = stats.stdev(data)
print("(6) Standard Deviation:", std_dev)
```
